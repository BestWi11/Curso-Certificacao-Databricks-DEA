## Relacionamentos

| Relação | Tipo | Descrição |
|---------|------|-----------|
| CUSTOMERS → ORDERS | 1:N | Um cliente pode ter vários pedidos |
| ORDERS → ORDER_ITEMS | 1:N | Um pedido contém vários itens |
| PRODUCTS → ORDER_ITEMS | 1:N | Um produto pode estar em vários itens |

---

## Estrutura das Camadas Medallion

```
📁 projeto-etl/
├── 📂 bronze/
│   ├── raw_customers
│   ├── raw_products
│   ├── raw_orders
│   └── raw_order_items
├── 📂 silver/
│   ├── cleaned_customers
│   ├── cleaned_products
│   ├── cleaned_orders
│   └── cleaned_order_items
└── 📂 gold/
    ├── vendas_por_cliente

```
